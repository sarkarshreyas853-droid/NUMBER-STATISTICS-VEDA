# Task 5: Number Statistics

This notebook identifies even and odd numbers in a list, then finds the largest and smallest numbers. It uses a loop, conditions and the modulo operator, and checks the results against the original data.

```
# Manually created numerical data
numbers = [14, 7, 23, 8, 41, 16, 3, 29, 12, 35]
print("Original data:", numbers)
```

## Separate even and odd numbers

A number is even when dividing it by 2 leaves no remainder, so `number % 2 == 0`. A loop checks each value and sorts it into one of two lists.

```
even_numbers = []
odd_numbers = []

for number in numbers:
    if number % 2 == 0:
        even_numbers.append(number)
    else:
        odd_numbers.append(number)

print("Even numbers:", even_numbers)
print("Odd numbers: ", odd_numbers)
```

## Largest and smallest numbers

The built-in `max()` and `min()` functions find the extremes.

```
largest = max(numbers)
smallest = min(numbers)

print(f"Largest number:  {largest}")
print(f"Smallest number: {smallest}")
```

## Check against the original data

The even and odd lists together must contain every number, and the largest and smallest values must match the original list.

```
assert len(even_numbers) + len(odd_numbers) == len(numbers)
assert all(n % 2 == 0 for n in even_numbers)
assert all(n % 2 == 1 for n in odd_numbers)
assert sorted(even_numbers + odd_numbers) == sorted(numbers)
assert largest == 41 and smallest == 3
print("All checks passed: every number is sorted once, and the extremes match the original data.")
```

## Answers to the interview questions

1. **How do you check whether a number is even?** Use the modulo operator: `number % 2 == 0` means the remainder is zero, so the number is even.
2. **What does the modulo operator do?** It returns the remainder after division. For example, `7 % 2` gives `1`, so 7 is odd.
3. **How can loops be used for data processing?** A loop applies the same check or calculation to every item in a collection, such as sorting numbers into groups, counting values, or summing a column.
